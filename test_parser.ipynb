```python
from pathlib import Path
from syncode.larkm import Lark
from syncode.parsers.incremental_parser import IncrementalParser

grammar_text = Path("c_grammar/c11.lark").read_text(encoding="utf-8")

base_parser = Lark(
    grammar_text,
    parser="lalr",
    lexer="basic",
    start="start",
    propagate_positions=True,
)

parser = IncrementalParser(base_parser)
result = parser.get_acceptable_next_terminals("int x;")

print("Parsed tokens:", [(t.type, str(t)) for t in parser.parsed_lexer_tokens])
print("Remainder:", repr(result.remainder))
print("State:", result.remainder_state)
```
